```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load raw dataset
# Adjust path if running from a different working directory
df = pd.read_csv("../data/raw/ai4i2020.csv")

# Failure distribution
sns.countplot(x="Machine failure", data=df)
plt.title("Machine Failure Distribution")
plt.show()

# Sensor trends before failure
df_fail = df[df["Machine failure"] == 1].head(200)
df_fail[["Air temperature [K]", "Process temperature [K]"]].plot()
plt.title("Sensor Trends Before Failure")
plt.show()
```
